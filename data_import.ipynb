```python
import pandas as pd
import numpy as np
from matplotlib import pyplot as plt
```

```python
def get_trace(df, x_col, y_col):
    mask = df[x_col].notna()
    x = df.loc[mask, x_col].values
    y    = df.loc[mask, y_col].values
    return x, y
```
